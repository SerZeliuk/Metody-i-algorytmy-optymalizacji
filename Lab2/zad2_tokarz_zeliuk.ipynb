# 1.1 Funkcja z zad.1

Three-hump camel function 

**f(vec)** - Funkcja zwraca wynik funkcji Three-hump camel function na podstawie podanego wektora argumentów. Tą funkcję będziemy chcieli zoptymalizować.

**f_grad(vec)** - Funkcja oblicza gradient, pochodne cząstkowe funkcji w dwóch kierunkach, zwraca listę dwuelementową, pierwszy element to pochodna cząstkowa po x, drugi po y

````julia

function f(vec)
    x, y = vec[1], vec[2]
    return 2 * x^2 - 1.05 * x^4 + x^6 / 6 + x*y + y^2
end

function f_grad(vec)
    x, y = vec[1], vec[2]
    return [4 * x - 4.2 * x^3 + x^5 + y, x + 2 * y]
end
````

# 1.4 Funkcje z ograniczeniami

Ponizej znajdują się funkcje, które zostaną zastosowane w optymalizacji z ograniczeniami:
- **g(vec)** - funkcja ograniczenia g w postaci paraboli 3D, przyjmuje listę z 2 wartościami i zwraca wyliczoną wartość. Wzór: g(x, y) = x^2 + y + b. Parametr b = 1.5 został dobrany tak, aby ograniczenie nie przechodziło przez żadne z minimów lokalnych. 
- **g_grad(vec)** - funkcja obliczająca pochodne cząstkowe po x i po y funckji ograniczenia g, przyjmuje listę 2 wartości i zwraca listę pochodnych cząstkowych.
- **p(vec)** - funkcja kary p, jest to wartość funkcji g podniesiona do kwadratu 
- **p_grad(vec)** - wylicza pochodną z funkcji p
- **fp(vec, ρ)** - funkcja kary fp, stosowana do obliczania wartości kary w zależności od wartości funkcji f. Parametry: vec - punkt o wartości x i y, ρ - parametr przy funkcji kary. Zwraca: wyliczoną wartość liczbową.
- **fp_grad(vec, ρ)** - gradient funkcji fp.

````julia
function g(vec)
    x, y = vec[1], vec[2]
    return x^2 + y + 1.5
end

function g_grad(vec)
    x, y = vec[1], vec[2]
    return [2 * x, 1]
end

function p(vec)
    return g(vec)^2
end

function p_grad(vec)
    return 2 * g(vec) * g_grad(vec)
end

function fp(vec, ρ)
    return f(vec) + ρ * p(vec)
end

function fp_grad(vec, ρ)
    return f_grad(vec) + ρ * p_grad(vec)
end
````

# 1.6 Metoda największego spadku ze stałym krokiem - funkcja kary, optymalizacja z ograniczeniem.

- **gradient_descent_with_penalty_method(fx, fx_grad, px, px_grad, fpx, fpx_grad, x0, k_max, α, ρ=1.0, γ=2.0, atol=1e-8, max_iter=100, tol=1e-6)** - Funkcja implementująca algorytm metody największego spadku wraz z funkcją kary. Zwraca znalezione minimum oraz wartości funkcji f i p dla kolejnych iteracji.

Argumenty:
- fx - funkcja optymalizowana
- fx_grad - gradient funkcji optymalizowanej
- px - funkcja kary
- px_grad - gradient funkcji kary
- fpx - funkcja kary w zależności od wartości funkcji optymalizowanej
- fpx_grad - gradient funkcji fpx
- x0 - punkt początkowy, od którego zaczynamy optymalizację
- k_max - maksmyalna liczba kroków dla ograniczenia
- α - wartość kroku, przez który mnożony jest gradient w metodzie największego spadku
- ρ=1.0 - początkowa wartość parametru przy funkcji kary
- γ=2.0 - parametr zwiększający parametr ρ w kolejnych iteracjach
- atol=1e-8 - współczynnik tolerancji dla metody kary
- max_iter=100 - maksymalna liczba iteracji dla metody optymalizacji największego spadku
- tol=1e-6 - współczynnik tolerancji dla metody największego spadku


````julia
function gradient_descent_with_penalty_method(fx, fx_grad, px, px_grad, fpx, fpx_grad, x0, k_max, α, ρ=1.0, γ=2.0, atol=1e-8, max_iter=100, tol=1e-6)
    x = x0
    θ = copy(x0)
    f_values = []
    p_values = []
    storage = zeros(length(θ))
    for k in 1 : k_max
        for i in 1:max_iter
            value_start = fpx(θ, ρ)
            storage = fpx_grad(θ, ρ)
            norm_storage = norm(storage)
            θ_new = θ - storage .* (α / norm_storage)
            value_stop = fpx(θ_new, ρ)
            push!(f_values, fx(θ))
            push!(p_values, px(θ))
            if abs(value_stop - value_start) < tol
                break
            else
                θ = θ_new
            end
        end
        x = θ
        ρ *= γ
        if p(x) < atol
            return x, f_values, p_values
        end
    end
    return x, f_values, p_values
end
````

# 1.8 Conjugate gradient descent with penalty method

````julia
function armijo_line_search(f, ∇f, x, d; α0=1.0, c=1e-4, τ=0.5)
    α = α0
    fx = f(x)
    gx = ∇f(x)
    slope = dot(gx, d)
    
    # Compute x + α*d inline (this allocates a new array each time)
    while f(x + α * d) > fx + c * α * slope
        α *= τ
        if α < 1e-14
            return 1e-14
        end
    end
    return α
end

# De‑optimized penalty-based nonlinear conjugate gradient method.
function penalty_ncg_method(
    f, f_grad,
    p, p_grad,
    fp, fp_grad,
    x0;
    k_max=10,
    ρ=1.0,
    γ=2.0,
    atol=1e-8,
    max_iter=50,
    tol=1e-6
)
    # Use a non-optimized copy: each update creates a new vector.
    x = copy(x0)
    
    # Use dynamically growing arrays (no preallocation)
    f_values = []
    p_values = []
    
    for k in 1:k_max
        r = -fp_grad(x, ρ)
        d = copy(r)
        
        for i in 1:max_iter
            nr = norm(r)
            if nr < tol
                break
            end
            
            α = armijo_line_search(
                (xx) -> fp(xx, ρ),
                (xx) -> fp_grad(xx, ρ),
                x, d;
                α0 = 0.3,
                c = 1e-4,
                τ = 0.5
            )
            
            # Instead of in-place update, create a new array each time.
            x = x + α * d
            
            push!(f_values, f(x))
            push!(p_values, p(x))
            
            r_new = -fp_grad(x, ρ)
            β = norm(r_new)^2 / nr^2
            
            if (i % 10) == 0
                d = copy(r_new)
            else
                # Compute new direction without in-place update.
                d = r_new + β * d
            end
            
            r = r_new
        end
        
        # Check if constraint is satisfied.
        if p(x) < atol
            return x, f_values, p_values
        end
        
        ρ *= γ
    end
    
    return x, f_values, p_values
end
````

````julia
for i in 1:10
    x0 = points[i]
    x_opt, fx_vals, px_vals = penalty_ncg_method(f, f_grad, p, p_grad, fp, fp_grad, x0; k_max=16,ρ=.4,γ=1.05,atol=1e-8,max_iter=100,tol=1e-6)
    plot(1:length(fx_vals), fx_vals, label="fx values for point $i")
    plot!(1:length(px_vals), px_vals, label="px values for point $i")
    display(plot!(title="Optimization from Point $i", xlabel="Iteration", ylabel="Function Value"))
end
````

# 1.9 Optymalizacja metody + benchmarking

W celu przyspieszenia działania funkcji, użyliśmy silnej typizacji oraz zmiany niektórych operacji na "in-place"


Na przykład:

```
f_values = Vector{Float64}(undef, total_iter)

p_values = Vector{Float64}(undef, total_iter)
```

Zamiast

```
f_values = []

p_values = []
```

Oraz 
```
@inbounds x .+= α .* d
```

Zamiast
```
x .= x .+ α .* d
```

````julia
function armijo_line_search_optimized(f, ∇f, x::Vector{Float64}, d::Vector{Float64};
                            α0::Float64 = 1.0, c::Float64 = 1e-4, τ::Float64 = 0.5)
    α = α0
    fx = f(x)
    gx = ∇f(x)
    slope = dot(gx, d)
    temp = similar(x)  

    while true
        @. temp = x + α * d
        if f(temp) <= fx + c * α * slope
            break
        end
        α *= τ
        if α < 1e-14
            return 1e-14
        end
    end
    return α
end

function penalty_ncg_method_optimized(f, f_grad, p, p_grad, fp, fp_grad,
                            x0::Vector{Float64};
                            k_max::Int = 10,
                            ρ::Float64 = 1.0,
                            γ::Float64 = 2.0,
                            atol::Float64 = 1e-8,
                            max_iter::Int = 50,
                            tol::Float64 = 1e-6)
    x = copy(x0)

    total_iter = k_max * max_iter
    f_values = Vector{Float64}(undef, total_iter)
    p_values = Vector{Float64}(undef, total_iter)
    idx = 0  

    for k in 1:k_max
        r = -fp_grad(x, ρ)
        d = copy(r)
        for i in 1:max_iter
            nr = norm(r)
            if nr < tol
                break
            end

            α = armijo_line_search_optimized(xx -> fp(xx, ρ), xx -> fp_grad(xx, ρ), x, d;
                                   α0 = 0.3, c = 1e-4, τ = 0.5)
            
            @inbounds x .+= α .* d

            idx += 1
            f_values[idx] = f(x)
            p_values[idx] = p(x)

            r_new = -fp_grad(x, ρ)
            β = (norm(r_new)^2) / (nr^2)
            
            if (i % 10) == 0
                d = copy(r_new)
            else
                @inbounds d .= r_new .+ β .* d
            end

            r = r_new
        end

        if p(x) < atol
            return x, view(f_values, 1:idx), view(p_values, 1:idx)
        end

        ρ *= γ
    end

    return x, view(f_values, 1:idx), view(p_values, 1:idx)
end
````
