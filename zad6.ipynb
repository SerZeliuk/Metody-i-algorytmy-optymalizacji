## Na 3.0

Do realizacji:

1. Weź implementację metody największego spadku z zadania 1. Użyj biblioteki Optuna do optymalizacji wielkości kroku gdzie kryterium jest średnia liczba iteracji wykonanych do uzyskania zbieżności dla 20 losowych punktów początkowych. W przypadku braku zbieżności ustaw dostatecznie wysoką karę. Użyj mechanizmu _early stopping_ w oparciu o raportowanie cząstkowych średnich. Kod można zorganizować w oparciu o Pythona wywołującego Julię (jak w przykładzie z PySR) lub odwrotnie (https://juliamanifolds.github.io/ManoptExamples.jl/stable/examples/HyperparameterOptimization/).
2. Zapisz w raporcie najlepszą wartość hiperparametru. Zaznacz w której próbie został uzyskany. Wykonaj minimum 500 prób przed przerwaniem optymalizacji hiperparametru.
